```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import xarray as xr

from aqua.core.graphics.util_timeseries import plot_timeseries_ensemble

rng = np.random.default_rng(0)
tm = pd.date_range("2000-01", periods=240, freq="MS")
ty = pd.date_range("2000", periods=20, freq="YS")


def make(t, std):
    """Synthetic ensemble mean and std."""
    mean = np.sin(np.arange(len(t)) / 6) + rng.normal(0, 0.1, len(t))
    return (
        xr.DataArray(mean, dims="time", coords={"time": t}),
        xr.DataArray(np.full(len(t), std), dims="time", coords={"time": t}),
    )


fig, ax = plt.subplots(figsize=(10, 4))
m, ms = make(tm, 0.2)
a, a_std = make(ty, 0.1)
plot_timeseries_ensemble(ax, m, "monthly mean", ms, kind="monthly")
plot_timeseries_ensemble(ax, a, "annual mean", a_std, kind="annual")
ax.legend()
plt.show()
```
